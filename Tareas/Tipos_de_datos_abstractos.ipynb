### Desafío 1: El misterio de las referencias (Gestión de Memoria)

**Problema:** Un estudiante intenta duplicar una lista de valores, pero al modificar la 'copia', la original también cambia. ¿Por qué sucede esto y cómo se arregla?

```python
# CÓDIGO CON ERROR
original_data = [10, 20, 30]

# El estudiante intenta copiar los datos
copy_data = original_data

copy_data.append(40)

print(f"Original: {original_data}")
print(f"Copia: {copy_data}")

# TODO para el alumno: Explicar por qué ambos tienen el 40
# y arreglar el código para que sean independientes.
```

Explicación del error:
En Python, las variables que almacenan listas no guardan los valores directamente, sino una referencia (la dirección de memoria) hacia donde está guardada esa lista.
Cuando el estudiante ejecutó copy_data = original_data, el programa no creó una lista nueva. Simplemente hizo que la variable copy_data apuntara a la misma dirección de memoria que original_data. Como ambas variables están conectadas a la misma lista física, al agregarle el 40 a una, ambas muestran el mismo cambio.

```python
# CÓDIGO CORREGIDO
original_data = [10, 20, 30]

# SOLUCIÓN: Usamos el método .copy() para crear un clon exacto,
# pero en una ubicación de memoria diferente e independiente.
copy_data = original_data.copy()

# Agregamos el 40 a la copia
copy_data.append(40)

# Comprobamos los resultados
print(f"Original: {original_data}")
print(f"Copia: {copy_data}")
```

### 1. Pila (Stack) implementada con un Vector
Una **Pila** es una estructura de datos abstracta que sigue el principio **LIFO** (*Last In, First Out* - Último en entrar, primero en salir).

**¿Cómo usa la estructura primitiva?**
Utilizamos un vector (una lista estándar de Python) como base. Para simular el comportamiento de la pila, solo permitimos agregar elementos al final de la lista (usando `.append()`) y retirar elementos también del final (usando `.pop()`).
*   **Analogía:** Una pila de platos en la cocina o el botón de "Deshacer" en tu editor de texto.

```python
class Stack:
    """Implementación de una Pila usando una lista."""

    def __init__(self):
        self.items = []  # Nuestro vector primitivo

    def push(self, item):
        """Agrega un elemento a la cima de la pila."""
        self.items.append(item)

    def pop(self):
        """Retira y devuelve el elemento de la cima."""
        if not self.is_empty():
            return self.items.pop()
        return None

    def is_empty(self):
        """Verifica si la pila está vacía."""
        return len(self.items) == 0

# Prueba en Colab
mi_pila = Stack()
mi_pila.push(5)
mi_pila.push(10)
print(f"Elemento retirado de la pila: {mi_pila.pop()}")  # Imprimirá 10
```

### 2. Cola (Queue) implementada con un Vector
Una **Cola** es una estructura de datos que sigue el principio **FIFO** (*First In, First Out* - Primero en entrar, primero en salir).

**¿Cómo usa la estructura primitiva?**
Al igual que la pila, usamos un vector (lista) de fondo. Sin embargo, la regla cambia: agregamos elementos al final de la lista (`.append()`), pero retiramos el elemento que está al principio de la lista (`.pop(0)`), simulando un avance secuencial.
*   **Analogía:** La fila para pagar en el supermercado o la cola de impresión de una impresora.

```python
class Queue:
    """Implementación de una Cola usando una lista."""

    def __init__(self):
        self.items = []  # Nuestro vector primitivo

    def enqueue(self, item):
        """Agrega un elemento al final de la cola."""
        self.items.append(item)

    def dequeue(self):
        """Retira y devuelve el primer elemento de la cola."""
        if not self.is_empty():
            # pop(0) saca el primer elemento de la lista
            return self.items.pop(0)
        return None

    def is_empty(self):
        return len(self.items) == 0

# Prueba en Colab
mi_cola = Queue()
mi_cola.enqueue("Cliente 1")
mi_cola.enqueue("Cliente 2")
print(f"Atendiendo a: {mi_cola.dequeue()}")  # Imprimirá "Cliente 1"
```

### 3. Grafo usando una Matriz de Adyacencia
Un **Grafo** es una estructura no lineal utilizada para representar relaciones (aristas) entre diferentes objetos (nodos o vértices).

**¿Cómo usa la estructura primitiva?**
Utilizamos una **Matriz** (una lista que contiene otras listas adentro, formando filas y columnas). Si tenemos 3 nodos, creamos una matriz de 3x3. Si el nodo 0 está conectado con el nodo 1, colocamos un `1` en las coordenadas `[0][1]` y `[1][0]`. Si no hay conexión, se mantiene en `0`.
*   **Analogía:** Las amistades en una red social o las rutas de vuelo entre ciudades.

```python
class Graph:
    """Grafo representado por una matriz cuadrada."""

    def __init__(self, num_nodes):
        self.num_nodes = num_nodes
        # Creamos una matriz llena de ceros (lista de listas)
        self.matrix = [[0] * num_nodes for _ in range(num_nodes)]

    def add_edge(self, node1, node2):
        """Crea una conexión bidireccional entre dos nodos."""
        self.matrix[node1][node2] = 1
        self.matrix[node2][node1] = 1

    def display(self):
        """Imprime la matriz de adyacencia."""
        for row in self.matrix:
            print(row)

# Prueba en Colab
mi_grafo = Graph(3)  # Grafo de 3 nodos (0, 1, 2)
mi_grafo.add_edge(0, 1)  # Conectamos el nodo 0 con el 1
print("Matriz de adyacencia del grafo:")
mi_grafo.display()
```

### 4. Tablero de Juego (Grid 2D) usando una Matriz
Los tableros bidimensionales son fundamentales en la programación de videojuegos y la representación de espacios cartesianos.

**¿Cómo usa la estructura primitiva?**
Nuevamente usamos una **Matriz** (lista de listas). Cada elemento de las listas internas representa una "casilla" del tablero. Mediante coordenadas (fila y columna), podemos acceder, modificar o verificar qué hay en cualquier punto exacto del mapa.
*   **Analogía:** Un juego de Tres en Raya (Tic-Tac-Toe), el mapa del Buscaminas o una hoja de cálculo en Excel.

```python
class GameBoard:
    """Tablero bidimensional usando una matriz."""

    def __init__(self, rows, cols):
        self.rows = rows
        self.cols = cols
        # Inicializamos la matriz con espacios vacíos
        self.grid = [["-"] * cols for _ in range(rows)]

    def place_token(self, row, col, token):
        """Coloca una ficha en las coordenadas dadas."""
        if 0 <= row < self.rows and 0 <= col < self.cols:
            self.grid[row][col] = token

    def show_board(self):
        """Muestra el tablero en consola."""
        for row in self.grid:
            # Unimos los elementos de la fila con un espacio
            print(" ".join(row))

# Prueba en Colab
tablero = GameBoard(3, 3)
tablero.place_token(1, 1, "X")
tablero.place_token(0, 2, "O")
print("Tablero de juego:")
tablero.show_board()
```

### 5. Árbol Binario Secuencial usando un Vector
Un **Árbol Binario** es una estructura jerárquica donde cada nodo tiene, como máximo, dos "hijos" (izquierdo y derecho). Aunque suelen construirse creando objetos que se enlazan entre sí, también pueden "aplanarse" matemáticamente.

**¿Cómo usa la estructura primitiva?**
Usamos un simple **Vector** (lista unidimensional). Aprovechamos una fórmula matemática para encontrar las posiciones: si un nodo está en el índice `i`, su hijo izquierdo forzosamente estará en el índice `(2*i) + 1` y su hijo derecho en `(2*i) + 2`.
*   **Analogía:** Un torneo de eliminación directa o la jerarquía de carpetas en tu computadora, pero "aplanados" en una sola fila.

```python
class VectorBinaryTree:
    """Árbol binario implementado sobre un vector (lista)."""

    def __init__(self, capacity):
        # Inicializamos el vector con valores Nulos (None)
        self.tree = [None] * capacity

    def set_root(self, value):
        """Establece la raíz del árbol en la posición 0."""
        self.tree[0] = value

    def set_left_child(self, parent_index, value):
        """Calcula el índice del hijo izquierdo y guarda el valor."""
        left_index = (2 * parent_index) + 1
        if left_index < len(self.tree):
            self.tree[left_index] = value

    def set_right_child(self, parent_index, value):
        """Calcula el índice del hijo derecho y guarda el valor."""
        right_index = (2 * parent_index) + 2
        if right_index < len(self.tree):
            self.tree[right_index] = value

# Prueba en Colab
arbol = VectorBinaryTree(7)
arbol.set_root("Raíz")
arbol.set_left_child(0, "Hijo Izq")
arbol.set_right_child(0, "Hijo Der")
print(f"Estructura del árbol en el vector: {arbol.tree}")
```
