**Implementación de una Lista Enlazada Simple**

Este código implementa una Lista Enlazada Simple, una estructura de datos dinámica fundamental en ciencias de la computación. A diferencia de las listas tradicionales (donde los datos se guardan en bloques continuos de memoria), una lista enlazada almacena cada dato en un objeto independiente llamado Nodo. Cada nodo guarda su valor y un "puntero" o referencia que lo conecta con el siguiente nodo de la cadena.

El código se divide en dos clases principales y cumple con las siguientes funciones:

**Clase Nodo:** Es el bloque de construcción básico. Guarda el dato (valor) y la referencia que apunta al siguiente elemento de la cadena (siguiente).

**Clase ListaSimple:** Es la clase administradora que gestiona la estructura a través de su punto de partida, conocido como cabeza (head).

Operaciones y responsabilidades de la clase:

**Control de lista vacía:** Utiliza el método esta_vacia() para verificar si la estructura tiene elementos antes de intentar realizar operaciones como eliminar o imprimir, previniendo así errores de ejecución.

**Inserción:** Permite agregar nuevos elementos tanto al principio de la estructura (insertar_al_inicio) como al final de la cadena (insertar_al_final), creando los nodos de forma dinámica solo cuando se necesitan.

**Búsqueda:** El método buscar() recorre la cadena secuencialmente, nodo por nodo, hasta encontrar el valor deseado, retornando un valor booleano (True o False).

**Eliminación por valor:** El método eliminar_por_valor() localiza un dato específico y lo elimina desenlazándolo de la cadena (haciendo que el nodo anterior apunte al nodo siguiente). Al quedar aislado, el recolector de basura de Python libera la memoria automáticamente.

**Recorrido visual:** El método imprimir() viaja a través de todos los punteros desde la cabeza hasta llegar al final (None), generando una representación gráfica en texto para visualizar el estado actual de los datos (ej. 10 -> 20 -> None).

```python
class Nodo:
    """Clase que representa un único elemento (nodo) en la memoria."""
    def __init__(self, valor):
        self.valor = valor
        self.siguiente = None  # Puntero al siguiente nodo


class ListaSimple:
    """Implementación de una Lista Enlazada Simple."""

    def __init__(self):
        # La lista comienza vacía, sin apuntar a ningún nodo
        self.cabeza = None

    def esta_vacia(self):
        """Control de lista vacía: retorna True si no hay elementos."""
        return self.cabeza is None

    def insertar_al_inicio(self, valor):
        """Inserta un nuevo nodo al principio de la lista."""
        nuevo_nodo = Nodo(valor)
        nuevo_nodo.siguiente = self.cabeza
        self.cabeza = nuevo_nodo

    def insertar_al_final(self, valor):
        """Inserta un nuevo nodo al final de la cadena."""
        nuevo_nodo = Nodo(valor)

        # Control de lista vacía
        if self.esta_vacia():
            self.cabeza = nuevo_nodo
            return

        actual = self.cabeza
        # Recorremos hasta llegar al último nodo
        while actual.siguiente is not None:
            actual = actual.siguiente

        actual.siguiente = nuevo_nodo

    def buscar(self, valor):
        """Busca un valor y retorna True si existe, False si no."""
        actual = self.cabeza
        while actual is not None:
            if actual.valor == valor:
                return True
            actual = actual.siguiente
        return False

    def eliminar_por_valor(self, valor):
        """Busca y elimina la primera aparición de un valor."""
        # Control de lista vacía
        if self.esta_vacia():
            print("Error: La lista está vacía, no se puede eliminar nada.")
            return False

        # Caso especial: Si el valor a eliminar es el primero (la cabeza)
        if self.cabeza.valor == valor:
            self.cabeza = self.cabeza.siguiente
            return True

        # Buscamos el nodo anterior al que queremos eliminar
        actual = self.cabeza
        while actual.siguiente is not None:
            if actual.siguiente.valor == valor:
                # "Saltamos" el nodo a eliminar para desenlazarlo
                actual.siguiente = actual.siguiente.siguiente
                return True
            actual = actual.siguiente

        print(f"El valor {valor} no se encontró en la lista.")
        return False

    def imprimir(self):
        """Recorre la lista e imprime sus elementos visualmente."""
        # Control de lista vacía
        if self.esta_vacia():
            print("Lista vacía: []")
            return

        actual = self.cabeza
        elementos = []
        while actual is not None:
            elementos.append(str(actual.valor))
            actual = actual.siguiente

        # Formato visual tipo: 10 -> 20 -> 30 -> None
        print(" -> ".join(elementos) + " -> None")


# ==========================================
# ZONA DE PRUEBAS
# ==========================================
mi_lista = ListaSimple()

print("1. Control de lista vacía inicial:")
mi_lista.imprimir()

print("\n2. Insertando elementos:")
mi_lista.insertar_al_final(10)
mi_lista.insertar_al_inicio(5)
mi_lista.insertar_al_final(20)
mi_lista.imprimir()

print("\n3. Búsqueda:")
print(f"¿Existe el 10?: {mi_lista.buscar(10)}")
print(f"¿Existe el 99?: {mi_lista.buscar(99)}")

print("\n4. Eliminando por valor:")
mi_lista.eliminar_por_valor(10) # Elimina del medio
mi_lista.imprimir()
mi_lista.eliminar_por_valor(5)  # Elimina la cabeza
mi_lista.imprimir()
```
