## **Estructura Dinámica basada en Tablas Hash (Diccionarios)**
Implementar el Tipo de Dato Abstracto "Conjunto" utilizando un diccionario de Python como estructura dinámica subyacente, con el fin de optimizar la eficiencia computacional. El ejercicio busca demostrar el principio de encapsulamiento en la Programación Orientada a Objetos: cambiar la estructura interna (de una lista a un diccionario) para acelerar drásticamente las operaciones de búsqueda, inserción y eliminación, manteniendo intacta la interfaz con la que interactúa el usuario final.

```python
class ConjuntoHash:
    """
    Implementación optimizada del ADT Conjunto.
    Utiliza un diccionario (Hash Map) como estructura dinámica subyacente
    para garantizar operaciones de búsqueda, inserción y borrado muy rápidas.
    """

    def __init__(self, elementos_iniciales=None):
        # La estructura interna ahora es un diccionario, no una lista
        self.elementos = {}

        if elementos_iniciales is not None:
            for elemento in elementos_iniciales:
                self.agregar(elemento)

    def contiene(self, elemento):
        """Verifica si un elemento pertenece al conjunto de forma instantánea."""
        # En un diccionario, el operador 'in' busca en las claves (muy rápido)
        return elemento in self.elementos

    def agregar(self, elemento):
        """Añade un nuevo elemento usándolo como clave en el diccionario."""
        if not self.contiene(elemento):
            # Guardamos el elemento como clave. El valor 'True' es solo relleno.
            self.elementos[elemento] = True

    def quitar(self, elemento):
        """Elimina un elemento del conjunto si este existe."""
        if self.contiene(elemento):
            del self.elementos[elemento]

    def union(self, otro_conjunto):
        """Retorna un nuevo ConjuntoHash con los elementos de ambos."""
        nuevo_conjunto = ConjuntoHash()

        # Iterar sobre un diccionario nos devuelve sus claves (nuestros elementos)
        for elemento in self.elementos:
            nuevo_conjunto.agregar(elemento)

        for elemento in otro_conjunto.elementos:
            nuevo_conjunto.agregar(elemento)

        return nuevo_conjunto

    def interseccion(self, otro_conjunto):
        """Retorna un nuevo ConjuntoHash con los elementos comunes a ambos."""
        nuevo_conjunto = ConjuntoHash()

        # Solo necesitamos iterar sobre el conjunto más pequeño para ser eficientes,
        # pero para mantenerlo simple, iteramos sobre el actual:
        for elemento in self.elementos:
            if otro_conjunto.contiene(elemento):
                nuevo_conjunto.agregar(elemento)

        return nuevo_conjunto

    def __str__(self):
        """Retorna una representación legible del conjunto."""
        if not self.elementos:
            return "{}"

        # Extraemos las claves del diccionario (que son nuestros datos reales)
        claves = self.elementos.keys()
        elementos_str = ", ".join(str(k) for k in claves)
        return f"{{{elementos_str}}}"


# ==========================================
# Zona de Pruebas
# ==========================================

print("--- Inicializando Conjuntos Optimizados ---")
conjunto_x = ConjuntoHash(["Manzana", "Pera", "Naranja"])
conjunto_y = ConjuntoHash(["Naranja", "Plátano", "Kiwi"])

print(f"Conjunto X: {conjunto_x}")
print(f"Conjunto Y: {conjunto_y}\n")

# Demostración de operaciones
print("--- Probando Operaciones ---")
print(f"¿X contiene 'Pera'?: {conjunto_x.contiene('Pera')}")
print(f"¿Y contiene 'Pera'?: {conjunto_y.contiene('Pera')}")

conjunto_union = conjunto_x.union(conjunto_y)
print(f"Unión (X U Y): {conjunto_union}")

conjunto_interseccion = conjunto_x.interseccion(conjunto_y)
print(f"Intersección (X ∩ Y): {conjunto_interseccion}")

# Probando agregar y quitar
print("\n--- Modificando Conjunto X ---")
conjunto_x.agregar("Uva")
conjunto_x.quitar("Manzana")
print(f"Conjunto X final: {conjunto_x}")
```

## **Sistema de Control de Estacionamiento Estático**
El código modela la lógica de un estacionamiento real con capacidad fija utilizando una estructura estática. En lugar de usar listas dinámicas que crecen indefinidamente, reserva un número exacto de "cajones" en la memoria al crearse y gestiona manualmente qué cajones están ocupados y cuáles están libres.

```python
class Estacionamiento:
    """
    Gestiona los lugares de un estacionamiento usando una estructura estática.
    """

    def __init__(self, capacidad_maxima):
        self.capacidad = capacidad_maxima
        # Arreglo estático que simula los cajones del estacionamiento
        self.cajones = [None] * self.capacidad

    def aparcar_vehiculo(self, matricula):
        """Asigna un cajón vacío a un vehículo ingresante."""
        # 1. Validamos que el vehículo no esté ya adentro
        if self.buscar_vehiculo(matricula) != -1:
            print(f"[!] Error: El vehículo con matrícula {matricula} ya está dentro.")
            return False

        # 2. Buscamos el primer espacio libre (None)
        for i in range(self.capacidad):
            if self.cajones[i] is None:
                self.cajones[i] = matricula
                print(f"[INGRESO] Vehículo {matricula} aparcado en el cajón {i + 1}.")
                return True

        # 3. Si terminamos el bucle y no hubo retorno, está lleno
        print(f"[RECHAZO] Estacionamiento lleno. El vehículo {matricula} debe esperar.")
        return False

    def retirar_vehiculo(self, matricula):
        """Libera el cajón ocupado por el vehículo que sale."""
        for i in range(self.capacidad):
            if self.cajones[i] == matricula:
                # Liberamos el espacio estático volviéndolo None
                self.cajones[i] = None
                print(f"[SALIDA] Vehículo {matricula} retirado. Cajón {i + 1} liberado.")
                return True

        print(f"[?] El vehículo {matricula} no se encuentra en las instalaciones.")
        return False

    def buscar_vehiculo(self, matricula):
        """Localiza un vehículo y retorna su número de cajón."""
        for i in range(self.capacidad):
            if self.cajones[i] == matricula:
                return i + 1  # Retornamos el índice + 1 por legibilidad
        return -1

    def contar_espacios_libres(self):
        """Realiza un cálculo matemático para saber la disponibilidad actual."""
        libres = 0
        for espacio in self.cajones:
            if espacio is None:
                libres += 1
        return libres

    def mostrar_mapa(self):
        """Muestra una representación visual del estacionamiento."""
        print("\n--- MAPA DEL ESTACIONAMIENTO ---")
        for i in range(self.capacidad):
            # Formato visual condicional
            estado = self.cajones[i] if self.cajones[i] is not None else "[ LIBRE ]"
            print(f"Cajón {i + 1}: {estado}")
        print(f"> Disponibilidad actual: {self.contar_espacios_libres()} de {self.capacidad}")
        print("--------------------------------\n")


# ==========================================
# EJEMPLO DE USO E INSTANCIACIÓN
# ==========================================

# 1. Instanciamos el estacionamiento comercial con 4 lugares
mi_parking = Estacionamiento(capacidad_maxima=4)

print(">>> ABRIENDO ESTACIONAMIENTO...")
mi_parking.mostrar_mapa()

# 2. Aparcamos vehículos
mi_parking.aparcar_vehiculo("ABC-123")
mi_parking.aparcar_vehiculo("XYZ-987")
mi_parking.aparcar_vehiculo("LMN-456")

# 3. Intentamos un error de lógica (Meter el mismo auto dos veces)
mi_parking.aparcar_vehiculo("ABC-123")

# 4. Llenamos el estacionamiento
mi_parking.aparcar_vehiculo("QWE-111")

# 5. Llega un auto cuando ya no hay lugar (Límite estático alcanzado)
mi_parking.aparcar_vehiculo("RTY-222")

# 6. Consultamos la información general
mi_parking.mostrar_mapa()

# 7. Un cliente sale, liberando el espacio, y entra el que estaba esperando
print("\n>>> UN CLIENTE SE RETIRA...")
mi_parking.retirar_vehiculo("XYZ-987")
mi_parking.aparcar_vehiculo("RTY-222")

# 8. Vemos el mapa por última vez
mi_parking.mostrar_mapa()
```
