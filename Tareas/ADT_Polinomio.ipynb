```python
class Polinomio:
    """
    Implementación del Tipo de Dato Abstracto Polinomio usando diccionarios.
    """
    def __init__(self):
        self.terminos = {}

    def agregar_termino(self, coef, exp):
        """Agrega un término. Si el exponente ya existe, suma los coeficientes."""
        if coef == 0:
            return  # No almacenamos términos nulos

        if exp in self.terminos:
            self.terminos[exp] += coef
            # Si al sumar se anula el término, lo eliminamos para limpiar memoria
            if self.terminos[exp] == 0:
                del self.terminos[exp]
        else:
            self.terminos[exp] = coef

    def evaluar(self, x):
        """Evalúa el polinomio reemplazando la 'x' por el valor numérico."""
        return sum(coef * (x ** exp) for exp, coef in self.terminos.items())

    def sumar(self, otro):
        """Suma el polinomio actual con otro y retorna uno nuevo."""
        resultado = Polinomio()
        for exp, coef in self.terminos.items():
            resultado.agregar_termino(coef, exp)
        for exp, coef in otro.terminos.items():
            resultado.agregar_termino(coef, exp)
        return resultado

    def restar(self, otro):
        """Resta al polinomio actual otro polinomio y retorna uno nuevo."""
        resultado = Polinomio()
        for exp, coef in self.terminos.items():
            resultado.agregar_termino(coef, exp)
        # Para restar, simplemente agregamos los términos del otro con signo invertido
        for exp, coef in otro.terminos.items():
            resultado.agregar_termino(-coef, exp)
        return resultado

    def derivar(self):
        """Calcula la primera derivada del polinomio y retorna uno nuevo."""
        resultado = Polinomio()
        for exp, coef in self.terminos.items():
            if exp > 0:
                # Regla de la cadena básica: cx^n -> (c*n)x^(n-1)
                resultado.agregar_termino(coef * exp, exp - 1)
        return resultado

    def __str__(self):
        """Genera una representación legible del polinomio."""
        if not self.terminos:
            return "0"

        terminos_ordenados = sorted(self.terminos.items(), reverse=True)
        text_terminos = []

        for exp, coef in terminos_ordenados:
            signo = "+" if coef > 0 else "-"
            coef_abs = abs(coef)

            # Formateo del coeficiente (omitimos el '1' si acompaña a una 'x')
            str_coef = "" if coef_abs == 1 and exp != 0 else str(coef_abs)

            # Formateo del exponente
            if exp == 0:
                termino = f"{coef_abs}" # Si es constante, mostramos el número normal
            elif exp == 1:
                termino = f"{str_coef}x"
            else:
                termino = f"{str_coef}x^{exp}"

            text_terminos.append(f"{signo} {termino}")

        # Unimos todo en un string
        resultado = " ".join(text_terminos)

        # Limpieza de signos iniciales para que se vea matemáticamente correcto
        if resultado.startswith("+ "):
            return resultado[2:]
        elif resultado.startswith("- "):
            return "-" + resultado[2:]
        return resultado


# ==========================================
# Zona de Pruebas (3 Polinomios Distintos)
# ==========================================

# Polinomio 1: 3x^2 + 5x - 2
p1 = Polinomio()
p1.agregar_termino(3, 2)
p1.agregar_termino(5, 1)
p1.agregar_termino(-2, 0)

# Polinomio 2: x^3 - 3x^2 + 4
p2 = Polinomio()
p2.agregar_termino(1, 3)
p2.agregar_termino(-3, 2)
p2.agregar_termino(4, 0)

# Polinomio 3: 2x^4 - x
p3 = Polinomio()
p3.agregar_termino(2, 4)
p3.agregar_termino(-1, 1)

print("--- REPRESENTACIÓN LEGIBLE ---")
print(f"P1(x) = {p1}")
print(f"P2(x) = {p2}")
print(f"P3(x) = {p3}\n")

print("--- OPERACIONES MATEMÁTICAS ---")
print(f"Suma (P1 + P2): {p1.sumar(p2)}")
print(f"Resta (P1 - P2): {p1.restar(p2)}")
print(f"Derivada de P1(x): {p1.derivar()}")
print(f"Derivada de P2(x): {p2.derivar()}")
print(f"Derivada de P3(x): {p3.derivar()}\n")

print("--- EVALUACIÓN DE VALORES ---")
x_val = 2
print(f"Evaluar P1(x) cuando x={x_val}: {p1.evaluar(x_val)}")
print(f"Evaluar P3(x) cuando x={x_val}: {p3.evaluar(x_val)}")
```
