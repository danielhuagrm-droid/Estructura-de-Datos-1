### Desafío 2: Implementación de TDA Pila (Stack)

**Problema:** La siguiente clase `Stack` tiene errores críticos. La función `pop` no valida si la pila está vacía y la lógica de `is_empty` es inversa. Además, el método `push` no está guardando los elementos correctamente.

```python
class Stack:
    def __init__(self):
        self.items = []

    def is_empty(self):
        # ERROR: Lógica invertida
        return len(self.items) > 0

    def push(self, item):
        # ERROR: No está agregando el item recibido
        self.items = [item]

    def pop(self):
        # ERROR: No verifica si hay elementos antes de borrar
        # y no retorna el elemento eliminado
        self.items.remove(-1)

# --- Prueba del Alumno ---
mi_pila = Stack()
mi_pila.push("A")
mi_pila.push("B")

print("¿Está vacía?", mi_pila.is_empty())
print("Elemento sacado:", mi_pila.pop())

# TODO: Corregir la clase Stack para que cumpla con el comportamiento LIFO
```

Análisis y corrección de errores en la clase Stack:

Lógica inversa en is_empty: El código original retornaba True si la longitud de la lista era mayor a 0 (len(self.items) > 0). Esto significa que decía que "estaba vacía" ¡justo cuando tenía elementos! Solución: Cambiarlo a len(self.items) == 0.

Reemplazo de datos en push: Al usar self.items = [item], el código estaba destruyendo la lista original y creando una nueva con un solo elemento cada vez que se agregaba algo. Solución: Usar el método .append(item) para ir agregando elementos al final del vector existente.

Uso incorrecto de remove en pop: El método remove(-1) busca literalmente el número -1 dentro de la lista para borrarlo, lo cual causó el error ValueError: list.remove(x): x not in list que vimos en la consola. Además, no devolvía el valor eliminado. Solución: Verificar primero si la pila no está vacía y luego usar el método .pop() de las listas de Python, el cual saca y retorna el último elemento automáticamente.

```python
class Stack:
    """Implementación de un TDA Pila (LIFO)."""

    def __init__(self):
        self.items = []

    def is_empty(self):
        # CORRECCIÓN 1: Retorna True solo si la longitud es exactamente 0
        return len(self.items) == 0

    def push(self, item):
        # CORRECCIÓN 2: Usamos .append() para agregar el elemento a la lista
        self.items.append(item)

    def pop(self):
        # CORRECCIÓN 3: Verificamos si no está vacía antes de sacar un elemento
        if not self.is_empty():
            # El método .pop() de la lista remueve y retorna el último elemento
            return self.items.pop()

        # Opcional: Avisar si intentan sacar algo de una pila vacía
        print("La pila está vacía, no hay elementos para sacar.")
        return None

# --- Prueba del Alumno ---
mi_pila = Stack()
mi_pila.push("A")
mi_pila.push("B")

print("¿Está vacía?", mi_pila.is_empty())
print("Elemento sacado:", mi_pila.pop())

# Comprobación extra para ver si quedó la "A"
print("Siguiente elemento sacado:", mi_pila.pop())
print("¿Está vacía al final?", mi_pila.is_empty())
```

### Desafío 3: Gestión de Memoria en el Heap (Nodos Dinámicos)

**Contexto:** En lenguajes de bajo nivel, el Heap requiere gestión manual. En Python es automático, pero conceptualmente, si perdemos la referencia al 'puntero' inicial, perdemos el acceso a los datos.

**Problema:** El siguiente código intenta crear una secuencia de dos nodos en el Heap, pero debido a una mala asignación, el segundo nodo reemplaza al primero en la variable de control, perdiendo la referencia al inicio de la estructura.

```python
class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.siguiente = None # Este es nuestro 'puntero' al siguiente espacio en el Heap

# --- Prueba del Alumno ---
# El estudiante intenta crear una cadena: [Nodo 1] -> [Nodo 2]

contenedor = Nodo("Datos Importantes 1")

# ERROR: El alumno intenta agregar el segundo nodo de esta forma:
contenedor = Nodo("Datos Importantes 2")

# Verificación
print(f"Contenido actual: {contenedor.valor}")
if contenedor.siguiente is None:
    print("ERROR: Se ha perdido la referencia al primer nodo. ¡Memory Leak conceptual!")

# TODO para el alumno: Corregir el código para que 'contenedor' mantenga el acceso al Nodo 1
# y el Nodo 1 apunte al Nodo 2.
```

Explicación del problema de gestión de memoria:
El error ocurre porque la variable contenedor funciona como una "etiqueta" que apunta a una dirección en la memoria (el Heap).

Al ejecutar contenedor = Nodo("Datos Importantes 2"), el estudiante no está agregando un nodo a la cadena, sino que está arrancando la etiqueta contenedor del primer nodo para pegársela al segundo. Al hacer esto, el primer nodo se queda "flotando" en la memoria sin ninguna variable que apunte a él, perdiéndose para siempre (el Recolector de Basura de Python lo eliminará eventualmente).

La solución:
En lugar de reasignar la variable principal contenedor, debemos usar el atributo siguiente del primer nodo para que funcione como un puente o enlace hacia el nuevo nodo que estamos creando.

```python
class Nodo:
    """Clase que representa un nodo básico para estructuras dinámicas."""

    def __init__(self, valor):
        self.valor = valor
        self.siguiente = None  # Puntero al siguiente espacio en el Heap


# --- Prueba del Alumno (CORREGIDA) ---
# 1. Creamos el primer nodo y guardamos su referencia en 'contenedor'
contenedor = Nodo("Datos Importantes 1")

# 2. SOLUCIÓN: En lugar de sobrescribir 'contenedor', asignamos el nuevo
# nodo al atributo 'siguiente' del nodo actual.
contenedor.siguiente = Nodo("Datos Importantes 2")


# --- Verificación ---
print(f"Contenido del primer nodo: {contenedor.valor}")

if contenedor.siguiente is None:
    print("ERROR: Se ha perdido la referencia al primer nodo. ¡Memory Leak conceptual!")
else:
    print(f"Contenido del segundo nodo: {contenedor.siguiente.valor}")
    print("¡Excelente! Los nodos están enlazados correctamente en el Heap.")
```

### Desafío 4: TDA Punto y Circunferencia (Paso por Referencia vs. Valor)

**Contexto:** Hemos definido un `Punto` y una `Circunferencia` que usa este `Punto` como su centro. En Python, los objetos se pasan por referencia. Esto puede llevar a comportamientos inesperados si no se maneja correctamente, simulando un problema donde se esperaría una copia por valor.

**Problema:** Un estudiante crea una circunferencia con un centro definido por un objeto `Punto`. Posteriormente, el estudiante modifica directamente las coordenadas del `Punto` original, esperando que la circunferencia mantenga su centro original. Sin embargo, debido a la naturaleza de las referencias, la circunferencia también cambia.

```python
class Punto:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

class Circunferencia:
    def __init__(self, centro, radio):
        self.centro = centro # ERROR: Se asigna la referencia directa al objeto Punto
        self.radio = radio

    def __str__(self):
        return f"Circunferencia con centro en {self.centro} y radio {self.radio}"

# --- Prueba del Alumno ---
mi_punto_original = Punto(1, 2)
mi_circunferencia = Circunferencia(mi_punto_original, 5)

print(f"Circunferencia inicial: {mi_circunferencia}")

# El alumno modifica el punto original
mi_punto_original.x = 10
mi_punto_original.y = 20

print(f"Circunferencia después de modificar el punto original: {mi_circunferencia}")

# TODO: Corregir la clase Circunferencia para que su centro no cambie si el Punto original es modificado.
# Es decir, la Circunferencia debe tener su propia copia de las coordenadas del centro.
```

Explicación: Referencias en Objetos
En Python, cuando pasamos un objeto (como una instancia de la clase Punto) a otra clase o función, no se pasa una copia de sus datos, sino una referencia a su ubicación en la memoria.

En el código original, al hacer self.centro = centro, la Circunferencia y la variable mi_punto_original estaban apuntando exactamente al mismo objeto Punto. Por eso, al cambiar las coordenadas del punto original a (10, 20), el centro de la circunferencia también reflejó ese cambio.

La solución:
Para lograr un comportamiento de "paso por valor" (independiente), debemos extraer las coordenadas x e y del punto recibido y utilizarlas para crear (instanciar) un nuevo objeto Punto que pertenezca exclusivamente a la Circunferencia.

```python
class Punto:
    """Representa una coordenada en un plano cartesiano 2D."""

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"


class Circunferencia:
    """Representa un círculo definido por un Punto central y un radio."""

    def __init__(self, centro, radio):
        # SOLUCIÓN: En lugar de guardar la referencia directa,
        # instanciamos un NUEVO objeto Punto copiando los valores x e y.
        self.centro = Punto(centro.x, centro.y)
        self.radio = radio

    def __str__(self):
        return f"Circunferencia con centro en {self.centro} y radio {self.radio}"


# --- Prueba del Alumno ---
mi_punto_original = Punto(1, 2)
mi_circunferencia = Circunferencia(mi_punto_original, 5)

print(f"Circunferencia inicial: {mi_circunferencia}")

# El alumno modifica el punto original
mi_punto_original.x = 10
mi_punto_original.y = 20

# Comprobación de que ahora son independientes
print(f"Circunferencia después de modificar el punto original: {mi_circunferencia}")
print(f"Estado del punto original: {mi_punto_original}")
```

### Desafío 5: TDA Automóvil y Garaje (Gestión de Objetos en Colecciones)

Contexto: Estamos modelando un Automovil y un Garaje que almacena una colección de automóviles. Cuando se trabaja con listas de objetos, es fácil cometer errores al añadir o modificar elementos, especialmente si no se comprende cómo Python maneja las referencias a objetos en el heap.

Problema: Un estudiante intenta añadir varios automóviles a un garaje. Sin embargo, debido a una reutilización incorrecta de una variable de objeto Automovil al añadir a la lista, termina con un garaje que parece contener el mismo automóvil repetido varias veces, en lugar de automóviles distintos.

```python
class Automovil:
    def __init__(self, marca, modelo, color):
        self.marca = marca
        self.modelo = modelo
        self.color = color

    def __str__(self):
        return f"{self.color} {self.marca} {self.modelo}"

class Garaje:
    def __init__(self):
        self.automoviles = []

    def agregar_automovil(self, auto):
        self.automoviles.append(auto)

    def mostrar_automoviles(self):
        print("Automóviles en el garaje:")
        for i, auto in enumerate(self.automoviles):
            print(f"  {i+1}. {auto}")

# --- Prueba del Alumno ---
mi_garaje = Garaje()

# El alumno intenta añadir dos coches diferentes
auto_temporal = Automovil("Toyota", "Corolla", "Rojo")
mi_garaje.agregar_automovil(auto_temporal)

# ERROR: Reutiliza la misma variable para 'crear' otro coche, pero en realidad solo modifica el objeto existente
auto_temporal.marca = "Honda"
auto_temporal.modelo = "Civic"
auto_temporal.color = "Azul"
mi_garaje.agregar_automovil(auto_temporal)

mi_garaje.mostrar_automoviles()

# TODO: Corregir el código para que el garaje contenga dos automóviles distintos con sus respectivas propiedades.
```

Explicación: Reutilización de Variables y Mutabilidad
El error ocurre porque la lista self.automoviles dentro del Garaje no guarda "copias" de los coches, sino referencias (enlaces) hacia los objetos en la memoria.

Al ejecutar auto_temporal.marca = "Honda", el estudiante no está creando un coche nuevo, sino que está modificando (mutando) el mismo coche que ya había metido al garaje. Luego, vuelve a meter una referencia a ese mismo objeto. Como resultado, la lista termina teniendo dos enlaces que apuntan exactamente al mismo coche modificado (el Honda Civic azul).

La solución:
Para tener dos coches distintos, es obligatorio instanciar un nuevo objeto llamando a la clase Automovil(...) por cada coche que queramos crear, asignándole un nuevo espacio independiente en la memoria.

```python
class Automovil:
    """Representa un vehículo con marca, modelo y color."""

    def __init__(self, marca, modelo, color):
        self.marca = marca
        self.modelo = modelo
        self.color = color

    def __str__(self):
        return f"{self.color} {self.marca} {self.modelo}"


class Garaje:
    """Gestiona una colección de objetos Automovil."""

    def __init__(self):
        self.automoviles = []

    def agregar_automovil(self, auto):
        self.automoviles.append(auto)

    def mostrar_automoviles(self):
        print("Automóviles en el garaje:")
        for i, auto in enumerate(self.automoviles):
            print(f"  {i+1}. {auto}")


# --- Prueba del Alumno CORREGIDA ---
mi_garaje = Garaje()

# 1. Instanciamos el primer objeto y lo agregamos
auto_toyota = Automovil("Toyota", "Corolla", "Rojo")
mi_garaje.agregar_automovil(auto_toyota)

# 2. SOLUCIÓN: Creamos una NUEVA instancia en lugar de reescribir la anterior.
# Esto reserva un nuevo espacio en la memoria (Heap).
auto_honda = Automovil("Honda", "Civic", "Azul")
mi_garaje.agregar_automovil(auto_honda)

# Opcional: También puedes instanciar y agregar directamente en una sola línea
# mi_garaje.agregar_automovil(Automovil("Ford", "Fiesta", "Negro"))

# Verificación
mi_garaje.mostrar_automoviles()
```

### Desafío 6: TDA Línea (Inmutabilidad Conceptual con Objetos Mutables)

**Contexto:** Una `Línea` se define por dos `Punto`s: un punto de inicio y un punto final. Conceptualmente, una vez que una línea se define, sus puntos extremos no deberían cambiar si se modifican los `Punto`s originales que se usaron para crearla, a menos que se quiera explícitamente.

**Problema:** Similar al desafío de la circunferencia, si la clase `Línea` almacena referencias directas a los objetos `Punto` mutables pasados durante su inicialización, cualquier modificación externa a esos `Punto`s afectará la `Línea` de forma inesperada. El estudiante espera que la línea permanezca estática después de su creación.

```python
class Punto:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

class Linea:
    def __init__(self, inicio, fin):
        self.inicio = inicio # ERROR: Almacena la referencia directa
        self.fin = fin       # ERROR: Almacena la referencia directa

    def __str__(self):
        return f"Línea de {self.inicio} a {self.fin}"

# --- Prueba del Alumno ---
punto_a = Punto(0, 0)
punto_b = Punto(5, 5)

mi_linea = Linea(punto_a, punto_b)
print(f"Línea original: {mi_linea}")

# El alumno modifica uno de los puntos originales
punto_a.x = 10
punto_a.y = 10

print(f"Línea después de modificar el punto 'A': {mi_linea}")

# TODO: Corregir la clase Linea para que sea "inmutable" conceptualmente;
# es decir, que los cambios en los objetos Punto externos no afecten la línea una vez creada.
```

Explicación: Inmutabilidad conceptual y Referencias
En Python, los objetos creados a partir de clases (como Punto) son mutables. Cuando pasamos punto_a y punto_b a la clase Linea, el código original simplemente guardaba las referencias a esos objetos en la memoria (self.inicio = inicio).

Al modificar punto_a.x fuera de la clase, el cambio se reflejaba dentro de la línea porque ambos apuntaban al mismo espacio físico en la memoria.

La solución:
Para lograr que la Linea sea "inmutable" frente a cambios externos, debemos aislarla. Lo hacemos extrayendo los valores x e y de los puntos recibidos y usándolos para instanciar dos nuevos objetos Punto completamente independientes que pertenezcan solo a la línea.

```python
class Punto:
    """Representa una coordenada en un plano cartesiano 2D."""

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"


class Linea:
    """Representa una línea definida por un punto de inicio y un punto final."""

    def __init__(self, inicio, fin):
        # SOLUCIÓN: Instanciamos nuevos objetos Punto usando las coordenadas
        # de los objetos originales para romper la referencia de memoria.
        self.inicio = Punto(inicio.x, inicio.y)
        self.fin = Punto(fin.x, fin.y)

    def __str__(self):
        return f"Línea de {self.inicio} a {self.fin}"


# --- Prueba del Alumno ---
punto_a = Punto(0, 0)
punto_b = Punto(5, 5)

mi_linea = Linea(punto_a, punto_b)
print(f"Línea original: {mi_linea}")

# El alumno modifica uno de los puntos originales externos
punto_a.x = 10
punto_a.y = 10

# Comprobación de la "inmutabilidad" de la línea
print(f"Línea después de modificar el punto 'A': {mi_linea}")
print(f"Estado del punto 'A' externo: {punto_a}")
```
