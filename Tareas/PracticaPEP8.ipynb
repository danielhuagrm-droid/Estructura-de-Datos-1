Definición del método para calcular el área de un triángulo, en el que se muestra el empleo de PEP 8.


```python
def calcular_area_triangulo(base, altura):
    """
    Calcula el área de un triángulo a partir de su base y altura.

    Parameters:
        base (int or float): La medida de la base del triángulo.
        altura (int or float): La medida de la altura del triángulo.

    Returns:
        int or float: El área calculada.
    """

    area = (base * altura) / 2
    return area

if __name__ == "__main__":
    # Llamada al método calcular_area_triangulo()
    mi_base = 10
    mi_altura = 5.5

    resultado = calcular_area_triangulo(mi_base, mi_altura)
    print(f"El área del triángulo es: {resultado}")
```

Definición de una clase para aplicar descuentos.

```python
class CalculadoraDescuentos:
    """
    Clase para aplicar descuentos a diferentes precios.
    """

    def __init__(self, porcentaje_descuento):
        """
        Inicializa la calculadora con un porcentaje fijo.

        Args:
            porcentaje_descuento (int or float): El descuento a aplicar (ej. 20 para 20%).
        """
        self.porcentaje_descuento = porcentaje_descuento

    def aplicar_descuento(self, precio_original):
        """
        Aplica el descuento configurado al precio proporcionado.

        Args:
            precio_original (int or float): El precio base del artículo.

        Returns:
            float: El precio final con el descuento ya aplicado.
        """
        descuento = precio_original * (self.porcentaje_descuento / 100)
        precio_final = precio_original - descuento
        return precio_final

if __name__ == "__main__":
    # Instanciamos la clase con un 15% de descuento
    mi_calculadora = CalculadoraDescuentos(15)

    precio_item = 200
    precio_rebajado = mi_calculadora.aplicar_descuento(precio_item)

    print(f"Precio original: ${precio_item}")
    print(f"Precio con descuento: ${precio_rebajado}")
```
